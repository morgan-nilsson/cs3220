```python

from data.mazeData import mazeData
from src.graphClass import Graph

mazeGraph = Graph(mazeData)
mazeGraph.nodes()
```

```python
from pyvis.network import Network

net = Network( 
    heading="Maze Graph",
    bgcolor ="#242020",
    font_color = "white",
    height = "750px",
    width = "100%"
)

net.add_nodes(mazeGraph.nodes())

nodeColors = {
    "start": "green",
    "end": "red",
    "deadend": "gray",
    "path": "blue",
}

for node in mazeGraph.nodes():
    if node == "start":
        net.get_node(node)["color"] = nodeColors["start"]
    elif node == "end":
        net.get_node(node)["color"] = nodeColors["end"]
    elif len(mazeGraph.get(node)) == 1:
        net.get_node(node)["color"] = nodeColors["deadend"]
    else:
        net.get_node(node)["color"] = nodeColors["path"]

# mazeGraph.graph_dict
net.nodes
```

```python
edges=[]
edges_labels=[]

for node_source in mazeGraph.nodes():
    for node_target, dist in mazeGraph.get(node_source).items():
        if set((node_source,node_target)) not in edges:
            net.add_edge(node_source,node_target, label=str(dist))
            edges.append(set((node_source,node_target)))
            edges_labels.append(str(dist))

net.show("mazeGraph.html")
            
```
