```python
from pyvis.network import Network
```

```python
net_sudoku = Network( heading="Lab6. Simple exam constraints",
                bgcolor ="#242020",
                font_color = "white",
                height = "750px",
                width = "100%"
) # do this
```

```python
nodeColors={
    "empty":"white",
    "filled": "yellow"
}
```

```python
examCSP.domains
```

```python
nodeColorsList=[]
nodeTitles=[]

for node in nodes:
    if len(examCSP.domains[node])==1:
        nodeColorsList.append(nodeColors["filled"])
        nodeTitles.append(str(examCSP.domains[node][0]))
    else:
        nodeColorsList.append(nodeColors["empty"])
        nodeTitles.append("")

nodeColorsList
```

```python
nodeTitles
```

```python
sizes=[10]*len(nodes)
sizes
```

```python
nodes=list(nodes)
nodes
```

```python
x_coords = []
y_coords = []

for node in nodes:
    if node.lower()=="a":
        x_coords.append(0)
        y_coords.append(0)
    elif node.lower()=="b":
        x_coords.append(-25)
        y_coords.append(50)
    elif node.lower()=="c":
        x_coords.append(25)
        y_coords.append(50)
    elif node.lower()=="d":
        x_coords.append(-35)
        y_coords.append(100)
    elif node.lower()=="e":
        x_coords.append(-15)
        y_coords.append(150)
    elif node.lower()=="f":
        x_coords.append(35)
        y_coords.append(100)
    elif node.lower()=="g":
        x_coords.append(45)
        y_coords.append(150)


    
x_coords,y_coords
```

```python
#hidden labels, and hover titles
net_sudoku.add_nodes(list(nodes), color=nodeColorsList,  size=sizes, title=nodeTitles,x=x_coords, y=y_coords)
```

```python
net_sudoku.nodes
```

```python
examCSP.neighbors
```

```python
for nodeFrom in examCSP.neighbors.keys():
    for nodeTo in examCSP.neighbors[nodeFrom]:
            net_sudoku.add_edge(nodeFrom,nodeTo, color="red")
```

```python
net_sudoku.edges
```

```python
net_sudoku.toggle_physics(False)
net_sudoku.show("Lab6. Exam CSP.html", notebook=False)
```
