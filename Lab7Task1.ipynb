**1. Formulate the problem (presented above) as a CSP problem.**

* The varibles would be the chairs.
* The neighbors would be the who chairs on your left and right.
* The domain is the all the people who can sit on the chair.
* The constraint is that no two chairs can have the person and that B can only be beside D OR NOTHING.

```python
from seatingCSP import seatingCSP
from src.utils import parse_neighbors
```

```python
neighbors = "1: 2 6; 2: 3; 3: 4; 4: 5; 5: 6; 6: "
varibles = ['1', '2', '3', '4', '5', '6']
domain = ['A', 'B', 'C', 'D', 'E', '_']
filled = {}

def seating_constraint(X, x, Y, y):

    # Forbidden adjacent pairs
    forbidden = {
        ('A', 'B'), ('B', 'A'),
        ('B', 'E'), ('E', 'B'),
        ('B', 'C'), ('C', 'B'),
    }

    return (x, y) not in forbidden
constraint = seating_constraint
```

```python
n = parse_neighbors(neighbors)
n
```

```python
seatingCSP = seatingCSP(varibles, neighbors ,domain, filled, constraint)
seatingCSP.neighbors
```

**3. Visualize the constraints graph with PyVis**

```python
from pyvis.network import Network
```

```python
nodes = seatingCSP.variables
nodes
```

```python
net_sudoku = Network( heading="Lab7. Seating CSP",
                bgcolor ="#1C1919",
                font_color = "white",
                height = "750px",
                width = "100%",
                directed= False
)
```

```python
nodeColors = {
    '1' : "red",
    '2' : "blue",
    '3' : 'green',
    '4' : 'yellow',
    '5' : 'orange',
    '6' : 'purple'
}
```

```python
seatingCSP.domains
```

```python
sizes=[10]*len(nodes)
sizes
```

```python
x_coords = []
y_coords = []

for node in nodes:
    if node.lower()=="1":
        x_coords.append(0)
        y_coords.append(0)
    elif node.lower()=="2":
        x_coords.append(50)
        y_coords.append(50)
    elif node.lower()=="3":
        x_coords.append(50)
        y_coords.append(125)
    elif node.lower()=="4":
        x_coords.append(0)
        y_coords.append(175)
    elif node.lower()=="5":
        x_coords.append(-50)
        y_coords.append(125)
    elif node.lower()=="6":
        x_coords.append(-50)
        y_coords.append(50)


    
x_coords,y_coords
```

```python

for i, node in enumerate(nodes):
    seat_domain = ""
    for j in seatingCSP.domains[node]:
        seat_domain += " " + str(j)
    net_sudoku.add_node(node, 
                        color=nodeColors[node], 
                        title = seat_domain,
                        size=sizes[i], 
                        x=x_coords[i], 
                        y=y_coords[i])
```

```python
net_sudoku.nodes
```

```python

for nodeFrom in seatingCSP.neighbors.keys():
    for nodeTo in seatingCSP.neighbors[nodeFrom]:
        net_sudoku.add_edge(nodeFrom, nodeTo, color="white")

net_sudoku.toggle_physics(False)
net_sudoku.show("Lab7_Seating_CSP.html", notebook=False)
```
